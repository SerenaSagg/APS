# Introducción

En este trabajo práctico se analiza la respuesta en frecuencia de distintos circuitos pasivos RLC. Para ello, se calcula la función de transferencia de cada circuito y se estudian su módulo y fase en función de la frecuencia.

A partir de los resultados obtenidos, se busca identificar el tipo de filtro correspondiente a cada circuito y determinar su ancho de banda.

# Teoría

## 1. Función de transferencia

La función de transferencia describe la relación entre la señal de salida y la señal de entrada de un circuito en el dominio de Laplace, considerando condiciones iniciales nulas.

Se define como:

$$
H(s)=\frac{V_2(s)}{V_1(s)}
$$

Donde:
- $V_2(s)$ es la tensión de salida.
- $V_1(s)$ es la tensión de entrada.
- $s$ es la variable compleja de Laplace.

Esta función permite analizar cómo responde un circuito ante distintas señales y determinar sus características como filtro.

## 2. Respuesta en frecuencia

La respuesta en frecuencia permite estudiar cómo varían la amplitud y la fase de la señal de salida en función de la frecuencia de entrada.

Se obtiene reemplazando:

$$
s=j\omega
$$

en la función de transferencia:

$$
H(j\omega)=\frac{V_2(j\omega)}{V_1(j\omega)}
$$

La respuesta en frecuencia se analiza mediante dos magnitudes:

**Módulo:** representa la relación entre las amplitudes de salida y entrada.

$$
|H(j\omega)|=\sqrt{\operatorname{Re}(H)^2+\operatorname{Im}(H)^2}
$$

**Fase:** representa el desfase entre la señal de salida y la de entrada.

$$
\phi(\omega)=\arg(H(j\omega))
$$

Si la función se expresa como un cociente, la fase puede calcularse restando los argumentos:

$$
\boxed{\phi(\omega)=\arg(N)-\arg(D)}
$$

Donde $N$ y $D$ son el numerador y el denominador de la función de transferencia.

## 3. Divisor de tensión

El divisor de tensión permite calcular la tensión sobre una impedancia cuando dos impedancias están conectadas en serie.

Su expresión general es:

$$
\boxed{V_2=V_1\frac{Z_2}{Z_1+Z_2}}
$$

Donde:
- $V_1$ es la tensión de entrada.
- $V_2$ es la tensión de salida sobre $Z_2$.
- $Z_1$ y $Z_2$ son las impedancias del circuito.

En el dominio de Laplace, las impedancias de los componentes pasivos ideales son:

**Resistencia:**

$$
Z_R=R
$$

**Inductor:**

$$
Z_L=sL
$$

**Capacitor:**

$$
Z_C=\frac{1}{sC}
$$

Cuando dos impedancias están conectadas en paralelo, su equivalente se calcula mediante:

$$
Z_{\mathrm{eq}}=\frac{Z_1Z_2}{Z_1+Z_2}
$$

Estas expresiones permiten simplificar los circuitos RLC y obtener sus funciones de transferencia.

## 4. Tipos de filtros

Los filtros son circuitos que permiten el paso de determinadas frecuencias y atenúan otras. Según su respuesta en frecuencia, pueden clasificarse en diferentes tipos.

### Filtro pasabajos

Permite el paso de frecuencias bajas y atenúa las frecuencias altas.

Su comportamiento característico es:

$$
|H(j0)|\neq 0
$$

$$
\lim_{\omega\to\infty}|H(j\omega)|=0
$$

En el trabajo práctico, este comportamiento corresponde al **circuito A**.

### Filtro pasaaltos

Atenúa las frecuencias bajas y permite el paso de frecuencias altas.

Su comportamiento característico es:

$$
|H(j0)|=0
$$

$$
\lim_{\omega\to\infty}|H(j\omega)|\neq 0
$$

En el trabajo práctico, este comportamiento corresponde al **circuito B**.

### Filtro pasabanda

Permite el paso de un intervalo de frecuencias intermedias y atenúa las frecuencias bajas y altas.

Su comportamiento característico es:

$$
|H(j0)|=0
$$

$$
|H(j\omega_0)|=H_{\max}
$$

$$
\lim_{\omega\to\infty}|H(j\omega)|=0
$$

Donde $\omega_0$ es la frecuencia central del filtro.

En el trabajo práctico, este comportamiento corresponde al **circuito C**.

## 5. Frecuencia de corte y ancho de banda

La frecuencia de corte es aquella en la que el módulo de la función de transferencia disminuye hasta un valor equivalente a $1/\sqrt{2}$ de la ganancia de referencia, lo que representa una reducción de $3\,\mathrm{dB}$.

$$
|H(j\omega_c)|=\frac{H_{\mathrm{ref}}}{\sqrt{2}}
$$

En los filtros pasabanda se consideran dos frecuencias de corte, una inferior y otra superior.

El ancho de banda se calcula como:

$$
\boxed{BW=\omega_{c2}-\omega_{c1}}
$$

Donde:
- $\omega_{c1}$ es la frecuencia de corte inferior.
- $\omega_{c2}$ es la frecuencia de corte superior.

El ancho de banda indica el intervalo de frecuencias que el filtro permite transmitir dentro de los límites establecidos.

## Análisis de circuitos RLC — Ejercicios A, B y C

**Datos utilizados en los tres ejercicios:**

$$
\omega_0=1\ \mathrm{rad/s},\qquad R=1\,\Omega,\qquad Q=\frac{\sqrt2}{2}
$$


## Ejercicio A — Salida sobre el capacitor

## 1. Impedancias y divisor de tensión

$$
Z_1=sL,\qquad Z_2=R,\qquad Z_3=\frac1{sC}
$$

$$
Z_{\mathrm{eq}}=Z_1+Z_2=sL+R
$$

$$
V_2=V_1\frac{\frac1{sC}}{\frac1{sC}+sL+R}
$$

$$
H(s)=\frac{V_2}{V_1}
=\frac{\frac1{sC}}{\frac1{sC}+sL+R}
=\boxed{\frac1{LCs^2+RCs+1}}
$$

## 2. Forma estándar y parámetros

Dividimos por $LC$:

$$
H(s)=\frac{\frac1{LC}}{s^2+\frac RL s+\frac1{LC}}
$$

Comparamos con:

$$
H(s)=\frac{\omega_0^2}{s^2+\frac{\omega_0}{Q}s+\omega_0^2}
$$

$$
\omega_0^2=\frac1{LC}
\quad\Longrightarrow\quad
\boxed{\omega_0=\frac1{\sqrt{LC}}}
$$

$$
\frac{\omega_0}{Q}=\frac RL
\quad\Longrightarrow\quad
Q=\frac{\omega_0 L}{R}
=\boxed{\frac1R\sqrt{\frac LC}}
$$

## 3. Cálculo de $L$ y $C$

$$
\omega_0=1
\quad\Longrightarrow\quad
LC=1
\quad\Longrightarrow\quad
L=\frac1C
$$

$$
\frac{\omega_0}{Q}=\frac RL
\quad\Longrightarrow\quad
\sqrt2=\frac1L
\quad\Longrightarrow\quad
\boxed{L=\frac1{\sqrt2}=\frac{\sqrt2}{2}\ }
$$

$$
C=\frac1L
\quad\Longrightarrow\quad
\boxed{C=\sqrt2\ }
$$

## 4. Función de transferencia

$$
\boxed{H_A(s)=\frac1{s^2+\sqrt2s+1}}
$$

Reemplazamos $s=j\omega$, con $j^2=-1$:

$$
H_A(j\omega)=\frac1{(j\omega)^2+j\sqrt2\omega+1}
=\boxed{\frac1{1-\omega^2+j\sqrt2\omega}}
$$

## 5. Módulo

$$
|H_A(j\omega)|
=\frac1{\sqrt{(1-\omega^2)^2+(\sqrt2\omega)^2}}
$$

$$
=\frac1{\sqrt{1-2\omega^2+\omega^4+2\omega^2}}
=\boxed{\frac1{\sqrt{1+\omega^4}}}
$$

$$
\omega=0\ \Longrightarrow\ |H_A|=1
$$

$$
\omega=1\ \Longrightarrow\ |H_A|=\frac1{\sqrt2}
$$

$$
\omega\to\infty\ \Longrightarrow\ |H_A|\to0
$$

## 6. Fase

$$
\phi_A(\omega)=\arg(N)-\arg(D)
$$

El numerador es $N=1$, así que $\arg(N)=0$.

El denominador es:

$$
D=1-\omega^2+j\sqrt2\omega
$$

**Si $0\le\omega<1$**, el denominador está en el primer cuadrante (en $\omega=0$, sobre el eje real positivo):

$$
\boxed{\phi_A(\omega)=-\arctan\left(\frac{\sqrt2\omega}{1-\omega^2}\right)}
$$

$$
\omega=0\ \Longrightarrow\ \boxed{\phi_A(0)=0}
$$

**Si $\omega=1$**, el denominador es $j\sqrt2$:

$$
\phi_A(1)=0-\frac\pi2
\quad\Longrightarrow\quad
\boxed{\phi_A(1)=-\frac\pi2}
$$

**Si $\omega>1$**, el denominador está en el segundo cuadrante:

$$
\phi_A(\omega)=0-\left[\pi-\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)\right]
$$

$$
\boxed{\phi_A(\omega)=-\pi+\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)}
$$

$$
\omega\to\infty\ \Longrightarrow\ \boxed{\phi_A\to-\pi}
$$

## 7. Cálculo del ancho de banda

La función de transferencia es:

$$
H_A(s)=\frac{1}{s^2+\sqrt{2}s+1}
$$

El módulo de la respuesta en frecuencia es:

$$
|H_A(j\omega)|=\frac{1}{\sqrt{1+\omega^4}}
$$

Para encontrar la frecuencia de corte, se utiliza la condición:

$$
|H_A(j\omega_c)|=\frac{1}{\sqrt{2}}
$$

Reemplazando:

$$
\frac{1}{\sqrt{1+\omega_c^4}}=\frac{1}{\sqrt{2}}
$$

Elevando al cuadrado:

$$
1+\omega_c^4=2
$$

$$
\omega_c^4=1
$$

Por lo tanto:

$$
\boxed{\omega_c=1\ \text{rad/s}}
$$

Como se trata de un filtro pasabajos, el ancho de banda es:

$$
BW_A=\omega_c-0
$$

$$
\boxed{BW_A=1\ \text{rad/s}}
$$

## Ejercicio B — Resistencia y capacitor en paralelo; salida sobre la inductancia

## 1. Impedancias y paralelo

$$
Z_R=R,\qquad Z_C=\frac1{sC},\qquad Z_L=sL
$$

$$
Z_{RC}=\frac{Z_RZ_C}{Z_R+Z_C}
=\frac{R\frac1{sC}}{R+\frac1{sC}}
=\boxed{\frac R{1+sRC}}
$$

## 2. Divisor de tensión

$$
V_2=V_1\frac{sL}{sL+\frac R{1+sRC}}
$$

$$
H(s)=\frac{V_2}{V_1}
=\frac{sL(1+sRC)}{sL(1+sRC)+R}
$$

$$
H(s)=\frac{sL+s^2RCL}{sL+s^2RCL+R}
$$

Dividimos numerador y denominador por $RCL$:

$$
\boxed{
H(s)=\frac{s^2+\frac{s}{RC}}{s^2+\frac{s}{RC}+\frac1{LC}}
}
$$

## 3. Forma estándar y parámetros

Comparamos el denominador con:

$$
s^2+\frac{\omega_0}{Q}s+\omega_0^2
$$

$$
\omega_0^2=\frac1{LC}
\quad\Longrightarrow\quad
\boxed{\omega_0=\frac1{\sqrt{LC}}}
$$

$$
\frac{\omega_0}{Q}=\frac1{RC}
\quad\Longrightarrow\quad
Q=\omega_0RC
=\boxed{R\sqrt{\frac CL}}
$$

## 4. Cálculo de $L$ y $C$

$$
Q=\omega_0RC
\quad\Longrightarrow\quad
\frac{\sqrt2}{2}=1\cdot1\cdot C
\quad\Longrightarrow\quad
\boxed{C=\frac{\sqrt2}{2}\ }
$$

$$
LC=1
\quad\Longrightarrow\quad
L=\frac1C
=\frac2{\sqrt2}
\quad\Longrightarrow\quad
\boxed{L=\sqrt2\ }
$$

## 5. Función de transferencia

$$
\boxed{H_B(s)=\frac{s^2+\sqrt2s}{s^2+\sqrt2s+1}}
$$

$$
H_B(j\omega)
=\frac{(j\omega)^2+j\sqrt2\omega}{(j\omega)^2+j\sqrt2\omega+1}
=\boxed{\frac{-\omega^2+j\sqrt2\omega}{1-\omega^2+j\sqrt2\omega}}
$$

## 6. Módulo

$$
|H_B(j\omega)|
=\frac{\sqrt{(-\omega^2)^2+(\sqrt2\omega)^2}}
{\sqrt{(1-\omega^2)^2+(\sqrt2\omega)^2}}
$$

$$
=\frac{\sqrt{\omega^4+2\omega^2}}
{\sqrt{1-2\omega^2+\omega^4+2\omega^2}}
=\boxed{\frac{\sqrt{\omega^4+2\omega^2}}{\sqrt{1+\omega^4}}}
$$

$$
\omega=0\ \Longrightarrow\ |H_B|=0
$$

$$
\omega=1\ \Longrightarrow\ |H_B|=\sqrt{\frac32}
$$

$$
\omega\to\infty\ \Longrightarrow\ |H_B|\to1
$$

## 7. Fase

$$
\phi_B(\omega)=\arg(N)-\arg(D)
$$

El numerador es:

$$
N=-\omega^2+j\sqrt2\omega
$$

Para $\omega>0$ está en el segundo cuadrante:

$$
\arg(N)=\pi-\arctan\left(\frac{\sqrt2\omega}{\omega^2}\right)
=\boxed{\pi-\arctan\left(\frac{\sqrt2}{\omega}\right)}
$$

El denominador es:

$$
D=1-\omega^2+j\sqrt2\omega
$$

**Si $0<\omega<1$**, el denominador está en el primer cuadrante:

$$
\arg(D)=\arctan\left(\frac{\sqrt2\omega}{1-\omega^2}\right)
$$

$$
\boxed{
\phi_B(\omega)=\pi-\arctan\left(\frac{\sqrt2}{\omega}\right)
-\arctan\left(\frac{\sqrt2\omega}{1-\omega^2}\right)
}
$$

$$
\omega\to0^+
\quad\Longrightarrow\quad
\phi_B\to\pi-\frac\pi2-0
=\boxed{\frac\pi2}
$$

**En $\omega=0$**, $H_B(j0)=0$, por lo que la fase está **indefinida**; $\pi/2$ es su límite por la derecha.

**Si $\omega=1$**, el denominador tiene ángulo $\pi/2$:

$$
\phi_B(1)=\pi-\arctan(\sqrt2)-\frac\pi2
$$

$$
\boxed{\phi_B(1)\approx0.6155\ \mathrm{rad}}
$$

**Si $\omega>1$**, el denominador está en el segundo cuadrante:

$$
\arg(D)=\pi-\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)
$$

$$
\phi_B(\omega)=
\pi-\arctan\left(\frac{\sqrt2}{\omega}\right)
-\left[\pi-\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)\right]
$$

$$
\boxed{
\phi_B(\omega)=
-\arctan\left(\frac{\sqrt2}{\omega}\right)
+\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)
}
$$

$$
\omega\to\infty
\quad\Longrightarrow\quad
\boxed{\phi_B\to0}
$$

## 8. Cálculo del ancho de banda

La función de transferencia es:

$$
H_B(s)=\frac{s^2+\sqrt{2}s}{s^2+\sqrt{2}s+1}
$$

El módulo de la respuesta en frecuencia es:

$$
|H_B(j\omega)|=
\frac{\sqrt{\omega^4+2\omega^2}}{\sqrt{1+\omega^4}}
$$

Como la ganancia tiende a 1 para frecuencias altas, se toma ese valor como referencia para determinar la frecuencia de corte:

$$
|H_B(j\omega_c)|=\frac{1}{\sqrt{2}}
$$

Reemplazando:

$$
\frac{\sqrt{\omega_c^4+2\omega_c^2}}{\sqrt{1+\omega_c^4}}
=\frac{1}{\sqrt{2}}
$$

Elevando al cuadrado:

$$
\frac{\omega_c^4+2\omega_c^2}{1+\omega_c^4}
=\frac{1}{2}
$$

Despejando:

$$
2\omega_c^4+4\omega_c^2=1+\omega_c^4
$$

$$
\omega_c^4+4\omega_c^2-1=0
$$

Realizando el cambio de variable:

$$
x=\omega_c^2
$$

Se obtiene:

$$
x^2+4x-1=0
$$

Aplicando la fórmula resolvente:

$$
x=\frac{-4\pm\sqrt{16+4}}{2}
$$

$$
x=-2\pm\sqrt{5}
$$

Como $x=\omega_c^2\geq 0$, se considera la solución positiva:

$$
\omega_c^2=\sqrt{5}-2
$$

Por lo tanto:

$$
\boxed{\omega_c=\sqrt{\sqrt{5}-2}\approx0.486\ \text{rad/s}}
$$

Como el filtro pasaaltos permite el paso de frecuencias superiores a la frecuencia de corte, sin presentar un límite superior en el modelo ideal:

$$
BW_B=\infty-\omega_c
$$

$$
\boxed{BW_B=\infty}
$$

## Ejercicio C — Inductancia y capacitor en paralelo; salida sobre el paralelo

## 1. Impedancias y paralelo

$$
Z_R=R,\qquad Z_L=sL,\qquad Z_C=\frac1{sC}
$$

$$
Z_{LC}=\frac{Z_LZ_C}{Z_L+Z_C}
=\frac{sL\frac1{sC}}{sL+\frac1{sC}}
=\boxed{\frac{sL}{1+s^2LC}}
$$

## 2. Divisor de tensión

$$
V_2=V_1\frac{Z_{LC}}{R+Z_{LC}}
=V_1\frac{\frac{sL}{1+s^2LC}}{R+\frac{sL}{1+s^2LC}}
$$

$$
H(s)=\frac{V_2}{V_1}
=\frac{sL}{R(1+s^2LC)+sL}
$$

$$
\boxed{H(s)=\frac{sL}{s^2RLC+sL+R}}
$$

Dividimos por $RLC$:

$$
\boxed{
H(s)=\frac{\frac{s}{RC}}{s^2+\frac{s}{RC}+\frac1{LC}}
}
$$

## 3. Forma estándar y parámetros

Comparamos el denominador con:

$$
s^2+\frac{\omega_0}{Q}s+\omega_0^2
$$

$$
\omega_0^2=\frac1{LC}
\quad\Longrightarrow\quad
\boxed{\omega_0=\frac1{\sqrt{LC}}}
$$

$$
\frac{\omega_0}{Q}=\frac1{RC}
\quad\Longrightarrow\quad
Q=\omega_0RC
=\boxed{R\sqrt{\frac CL}}
$$

## 4. Cálculo de $L$ y $C$

$$
C=\frac{Q}{\omega_0R}
=\frac{\sqrt2/2}{1\cdot1}
\quad\Longrightarrow\quad
\boxed{C=\frac{\sqrt2}{2}\ }
$$

$$
LC=1
\quad\Longrightarrow\quad
L=\frac1C=\frac2{\sqrt2}
\quad\Longrightarrow\quad
\boxed{L=\sqrt2\ }
$$

## 5. Función de transferencia

$$
\boxed{H_C(s)=\frac{\sqrt2s}{s^2+\sqrt2s+1}}
$$

$$
H_C(j\omega)=\frac{j\sqrt2\omega}{(j\omega)^2+j\sqrt2\omega+1}
=\boxed{\frac{j\sqrt2\omega}{1-\omega^2+j\sqrt2\omega}}
$$

## 6. Módulo

Para $\omega\ge0$:

$$
|H_C(j\omega)|=
\frac{\sqrt2\omega}{\sqrt{(1-\omega^2)^2+(\sqrt2\omega)^2}}
$$

$$
=\frac{\sqrt2\omega}{\sqrt{1-2\omega^2+\omega^4+2\omega^2}}
=\boxed{\frac{\sqrt2\omega}{\sqrt{1+\omega^4}}}
$$

$$
\omega=0\ \Longrightarrow\ |H_C|=0
$$

$$
\omega=1\ \Longrightarrow\ |H_C|=1
$$

$$
\omega\to\infty\ \Longrightarrow\ |H_C|\to0
$$

## 7. Fase

$$
\phi_C(\omega)=\arg(N)-\arg(D)
$$

El numerador es $N=j\sqrt2\omega$, de modo que, para $\omega>0$:

$$
\boxed{\arg(N)=\frac\pi2}
$$

El denominador es:

$$
D=1-\omega^2+j\sqrt2\omega
$$

**Si $0<\omega<1$**, el denominador está en el primer cuadrante:

$$
\boxed{
\phi_C(\omega)=\frac\pi2-\arctan\left(\frac{\sqrt2\omega}{1-\omega^2}\right)
}
$$

$$
\omega\to0^+
\quad\Longrightarrow\quad
\boxed{\phi_C\to\frac\pi2}
$$

**En $\omega=0$**, $H_C(j0)=0$, así que la fase está **indefinida**; $\pi/2$ es su límite por la derecha.

**Si $\omega=1$**, numerador y denominador tienen ángulo $\pi/2$:

$$
\phi_C(1)=\frac\pi2-\frac\pi2
\quad\Longrightarrow\quad
\boxed{\phi_C(1)=0}
$$

**Si $\omega>1$**, el denominador está en el segundo cuadrante:

$$
\phi_C(\omega)=\frac\pi2-
\left[\pi-\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)\right]
$$

$$
\boxed{
\phi_C(\omega)=-\frac\pi2+
\arctan\left(\frac{\sqrt2\omega}{\omega^2-1}\right)
}
$$

$$
\omega\to\infty
\quad\Longrightarrow\quad
\boxed{\phi_C\to-\frac\pi2}
$$

## 8. Cálculo del ancho de banda

La función de transferencia es:

$$
H_C(s)=\frac{\sqrt{2}s}{s^2+\sqrt{2}s+1}
$$

El módulo de la respuesta en frecuencia es:

$$
|H_C(j\omega)|=
\frac{\sqrt{2}\omega}{\sqrt{1+\omega^4}}
$$

Para determinar las frecuencias de corte, se utiliza:

$$
|H_C(j\omega_c)|=\frac{1}{\sqrt{2}}
$$

Reemplazando:

$$
\frac{\sqrt{2}\omega_c}{\sqrt{1+\omega_c^4}}
=\frac{1}{\sqrt{2}}
$$

Elevando al cuadrado:

$$
\frac{2\omega_c^2}{1+\omega_c^4}=\frac{1}{2}
$$

Despejando:

$$
4\omega_c^2=1+\omega_c^4
$$

$$
\omega_c^4-4\omega_c^2+1=0
$$

Realizando el cambio de variable:

$$
x=\omega_c^2
$$

Se obtiene:

$$
x^2-4x+1=0
$$

Aplicando la fórmula resolvente:

$$
x=\frac{4\pm\sqrt{16-4}}{2}
$$

$$
x=2\pm\sqrt{3}
$$

Por lo tanto, las frecuencias de corte son:

$$
\boxed{\omega_{c1}=\sqrt{2-\sqrt{3}}\approx0.518\ \text{rad/s}}
$$

$$
\boxed{\omega_{c2}=\sqrt{2+\sqrt{3}}\approx1.932\ \text{rad/s}}
$$

El ancho de banda de un filtro pasabanda se calcula como:

$$
BW_C=\omega_{c2}-\omega_{c1}
$$

Reemplazando:

$$
BW_C=\sqrt{2+\sqrt{3}}-\sqrt{2-\sqrt{3}}
$$

$$
BW_C\approx1.932-0.518
$$

Por lo tanto:

$$
\boxed{BW_C=\sqrt{2}\approx1.414\ \text{rad/s}}
$$

En este trabajo práctico se analizaron las funciones de transferencia y las respuestas en frecuencia de tres circuitos pasivos RLC.

A partir del estudio del módulo y la fase, se identificaron los circuitos como filtros pasabajos, pasaaltos y pasabanda. Además, se determinaron sus frecuencias de corte y anchos de banda, verificando cómo la configuración de los componentes influye en el comportamiento de cada filtro.
